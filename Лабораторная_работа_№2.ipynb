## **Задание**

Необходимо вычислить значение выражения:

$$
p = 1 - \frac{\cos(X)\cdot\sin^2(Y)}{1\cdot 2} + \frac{\cos^4(X)\cdot\sin^3(Y)}{3\cdot 4} - \frac{\cos^5(X)\cdot\sin^6(Y)}{5\cdot 6} + \ldots
$$

Для каждой переменной необходимо вывести ее значение и тип.

## **Анализ выражения и определение переменных**

Проанализируем выражение и определим переменные, которые потребуются для выполнения задания. Независимые переменные: $X$ и $Y$ Переменная, которая должна быть вычислена на основе независимых: $p$.

Следует обратить внимание, что в выражении неизвестное число слагаемых, поэтому необходимо ввести переменную $N$ для представления количества слагаемых.

Обратите внимание, что знаки слагаемых чередуются, а степени числителя и множетели знаменателя привязаны к порядковому номеру слагаемого.

## **Вычисление выражение, печать результатов**

Код для вычисления значения выражения $p$:

````python
import math

X=5
Y=10
N=7
p = 1

for i in range(1, N + 1):
    # Степени числителей
    if i % 2 == 1:
        stepen_cos = i
        stepen_sin = i + 1
    else:
        stepen_cos = i + 2
        stepen_sin = i + 1

    chis = math.cos(X) ** stepen_cos * math.sin(Y) ** stepen_sin
    znam = (2 * i - 1) * (2 * i)

    if i % 2 == 1:
        p -= chis / znam
    else:
        p += chis / znam

print('X =', X, 'тип:', type(X))
print('Y =', Y, 'тип:', type(Y))
print('N =', N, 'тип:', type(N))
print('p =', p, 'тип:', type(p))
````
