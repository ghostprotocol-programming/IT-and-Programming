## **Решение**

### **Объявление функции**

```python
def numbers(N):
   for i in range(-N, N+1):
      if i%4==0 or i%5==0 or i%7==0:
         print(i)
```

### **Демонстрация использования функции**

```python
numbers(20)
```

## **Ответы на контрольные вопросы:**

1. Функция с возвратом значения выполняет действие и возвращает результат, который можно использовать дальше в программе. Функция без возврата значения только выполняет действие и не возвращает результат
2. Библиотечная функция - это готовая функция, которая уже написана и хранится в библиотеке языка программирования. Её можно использовать в своей программе
3. Библиотечные функции похожи на «чёрные ящики», потому что программисту не обязательно знать, как они устроены внутри. Достаточно знать, какие данные передать функции и какой результат она выдаст
4. x = random.randint(1, 100) случайным образом выбирает целое число от 1 до 100 включительно и записывает его в переменную x
5. print(random.randint(1, 20)) - случайным образом выбирает целое число от 1 до 20 включительно и выводит его на экран
6. random.seed(20) - устанавливает начальное значение (зерно) генератора случайных чисел. Благодаря этому при одинаковом seed последовательность случайных чисел будет одинаковой
7. Функции позволяют написать код один раз и многократно использовать его в разных местах программы, вызывая функцию по имени. Это уменьшает повторение кода и упрощает его исправление
8. Две основные части определения функции:
Заголовок функции - содержит ключевое слово def, имя функции и её параметры.
Тело функции - блок команд, которые выполняются при вызове функции
9. Параметр по умолчанию задаётся при определении функции с помощью знака =
10. Рекурсия - это способ, при котором функция вызывает саму себя. Чтобы рекурсия не продолжалась бесконечно, в функции должно быть условие остановки (базовый случай)
