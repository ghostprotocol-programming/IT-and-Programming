## **Задание**

Необходимо вычислить значение выражения:

**$$
Se = w111 + bt - x + y \cdot w
$$**

Для каждой переменной необходимо вывести ее значение и тип.

## **Анализ выражения и определение переменных**

Проанализируем выражение и определим переменные, которые потребуются для выполнения задания. Независимые переменные: $w111, bt, x, y, w$.
Переменная, которая должна быть вычислена на основе независимых: $Se$.

```python
w111=10
bt=5
x=3
y=4
w=6

print("w111 =", w111, ", type(w111) =", type(w111))
print("bt =", bt, ", type(bt) =", type(bt))
print("x =", x, ", type(x) =", type(x))
print("y =", y, ", type(y) =", type(y))
print("w =", w, ", type(w) =", type(w))
```

## **Вычисление выражение, печать результатов**

Код для вычисления значения выражения $Se$:

```python
Se=w111+bt-x+y*w

print('Se=',Se)
print('Type of Se:', type(Se))
```
