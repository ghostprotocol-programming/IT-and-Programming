## **Задание**

В файле data_v2.txt содержится последовательность символов, содержащая буквы латинского алфавита и десятичные цифры. Необходимо посчитать количество символов, расположенных между заглавными буквами.

## **Анализ задания, получение данных**

Файл data_v2.txt расположен репозитории. Выполним копирование файла в локальное хранилище Google Colab, используя следующую команду:

```python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/main/data/data_v2.txt
```

Реализуем чтение всего файла в строку с использованием метоа `open()` и `read()`:

```python
f=open('data_v2.txt')
s=f.read().strip()
print(len(s))
```

## **Обработка символьной последовательности**

Строки, подобно списка, являются коллекциями. Только в качестве элементов последовательности выступают символы. Для строк также используются функции нахождения длины `len()`, операция индексации `[]`.

## **Решение**

```python
count=0
target=0
flag=False

for c in s:
  if 'A'<=c<='Z':
    if flag:
      target+=count
    flag=True
    count=0
  else:
    if flag:
      count+=1
print('Количество символов между заглавными буквами = %d' % target)
```

## **Ответы на контрольные вопросы:**

1. Функция `read()` считывает весь файл или указанное количество символов и возвращает их в виде строки, `readline()` считывает одну строку из файла и возвращает её в виде строки, `readlines()` считывает все строки файла и возвращает их в виде списка строк. Все три функции используются для чтения содержимого открытого текстового файла, но отличаются объёмом и форматом возвращаемых данных.
2. С помощью цикла `for`, перебирая файл построчно
3. Закрытие файла необходимо для освобождения системных ресурсов, связанных с открытым файлом, и завершения работы с ним. При записи закрытие также гарантирует сохранение накопленных данных в файле
4. Позиция считывания файла - это текущее место в файле, с которого будет выполняться следующая операция чтения. При первоначальном открытии файла в режиме чтения позиция считывания находится в самом начале файла
5. Файл открывается в режиме `a` (дозапись). При этом существующее содержимое файла сохраняется, а новые данные записываются в конец файла
6. Исключение - это ошибка или особая ситуация, возникающая во время выполнения программы и нарушающая обычный порядок её выполнения. Исключение можно обработать с помощью конструкции `try`/`except`
7. Если возникшее исключение не обрабатывается с помощью `try`/`except`, выполнение программы прекращается, а Python выводит сообщение об ошибке и информацию о месте, где она произошла
8. Возникает исключение `FileNotFoundError`
9. Возникает исключение `ValueError`
10. При открытии несуществующего файла в режиме дозаписи `a` файл автоматически создаётся. При этом, если файл уже существует, его содержимое не удаляется, а новые данные добавляются в конец файла
