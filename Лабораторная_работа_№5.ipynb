## **Задание**

В файле pr5.txt содержится последовательность целых чисел. Элементы последовательности могут принимать целые значения от –10 000 до 10 000 включительно. Определите количество троек, в которых хотя бы два из трёх элементов меньше, чем среднее арифметическое всех чисел в файле, и десятичная запись всех трёх элементов тройки содержит цифру 1. В ответе запишите два числа: сначала количество найденных троек, а затем – максимальную сумму элементов таких троек. В данной задаче под тройкой подразумевается три идущих подряд элемента последовательности.

## **Решение**

Файл pr5.txt расположен репозитории. Выполним копирование файла в локальное хранилище Google Colab, используя следующую команду:

```python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/refs/heads/main/data/pract5/pr5.txt
```

Реализуем чтение строк из файла и конвертацию отдельных строк в числовые значения. Каждое получаемое число будем добавлять в список `data`.

```python
# Создаем пустой список
data = []
# Выполняем операции чтения и заполнение списка
with open('pr5.txt') as F:
  for line in F:
    digit = int(line)
    data.append(digit)

# Посмотрим, что мы считали...
print(len(data))
print(data[:10])
```

Теперь ответим на поставленные вопросы используя список `data`

```python
# Вычислим среднее арифметическое всех чисел в файле
sr = sum(data) / len(data)
print(sr)
```

```python
# Подсчет количества подходящих троек
counter = 0
max_sum = None

for i in range(len(data) - 2):
    # Три идущих подряд элемента
    x, y, z = data[i], data[i + 1], data[i + 2]
    triple = [x, y, z]

    # Хотя бы два элемента меньше среднего арифметического,
    # и запись каждого из трех чисел содержит цифру 1
    if sum(number < sr for number in triple) >= 2 and all('1' in str(number) for number in triple):
        counter += 1
        triple_sum = sum(triple)
        if max_sum is None or triple_sum > max_sum:
            max_sum = triple_sum

print(f'Количество подходящих троек = {counter}')
print(f'Максимальная сумма = {max_sum}')
```
