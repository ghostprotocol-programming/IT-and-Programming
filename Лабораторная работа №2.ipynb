````python
import math

X = float(input('Введите X: '))
Y = float(input('Введите Y: '))
N = int(input('Введите количество слагаемых N: '))

p = 1

for i in range(1, N + 1):
    # Степени числителей
    if i % 2 == 1:
        stepen_cos = i
        stepen_sin = i + 1
    else:
        stepen_cos = i + 2
        stepen_sin = i + 1

    chis = math.cos(X) ** stepen_cos * math.sin(Y) ** stepen_sin
    znam = (2 * i - 1) * (2 * i)

    if i % 2 == 1:
        p -= chis / znam
    else:
        p += chis / znam

print('X =', X, 'тип:', type(X))
print('Y =', Y, 'тип:', type(Y))
print('N =', N, 'тип:', type(N))
print('p =', p, 'тип:', type(p))
````
