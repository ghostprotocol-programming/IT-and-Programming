Теперь ответим на поставленные вопросы, используя строку `mystr`

```python
mystr = open('pr6_1.txt').read().strip()

counter = 0

for i in range(len(mystr)-2):
  if mystr[i] == 'O' and mystr[i+1] in '0123456789' and mystr[i+2] in '0123456789':
    counter += 1

print(counter)
```
